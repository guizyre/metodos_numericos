# **1)Implemente o algoritmo da iteração de ponto fixo usando somente a biblioteca Numpy.**

```python
import numpy as np

def ponto_fixo(g, x0, tot=10**-4, max_itera=15):
  x = np.asarray(x0, dtype=float)

  for k in range(max_itera):
    novo_x = g(x)
    print(f"Interação {k+1}: {novo_x}")
    if np.all(np.abs(novo_x - x) < tot):
      return novo_x
    x = novo_x
  print(f"Erro de convergência: O método não atingiu a tolerância em {max_itera} iterações.")
  return None
```

# **2)Use a iteração de ponto fixo simples para localizar a raiz de f(x) = 2sin(raiz(x)− x, tendo x0 = 0, 5 e adotando como critério de parada o erro ea ≤ 0, 001%.**

```python
import numpy as np
def g(x):
  return 2 * np.sin(np.sqrt(x))
raiz = ponto_fixo(g, x0=0.5, tot=10**-5)
print("A raiz encontrada é:", raiz)


```
