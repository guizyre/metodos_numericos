```python
import numpy as np
import scipy.linalg as sla

def escalona(A, b):
    Ab = np.c_[A.astype(float), b.astype(float)]
    n = len(A)
    for c_idx in range(n - 1):
        for l_idx in range(c_idx + 1, n):
            m = Ab[l_idx, c_idx] / Ab[c_idx, c_idx]
            Ab[l_idx] = Ab[l_idx] - m * Ab[c_idx]
    return Ab[:, :-1], Ab[:, -1]

def substituicao_retroativa(U, c):
    n = len(c)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (c[i] - np.dot(U[i, i + 1:], x[i + 1:])) / U[i, i]
    return x

def resolver_sistema_gauss(A, b):
    U, c = escalona(A, b)
    x = substituicao_retroativa(U, c)
    return x, U, c
```

## Exercícios de Fixação 1 e 2

### Fixação 1
* **Sistemas Triviais:** Matrizes diagonais ou triangulares onde o isolamento das incógnitas é direto (substituição progressiva ou regressiva).
* **Sistemas Não Triviais:** Coeficientes dispersos fora da diagonal com acoplamento mútuo simultâneo entre variáveis, demandando métodos diretos de escalonamento ou métodos iterativos.

### Fixação 2
* **Eliminação de Gauss:** Transforma $Ax=b$ em $Ux=c$, alterando a matriz e o vetor independente.
* **Decomposição LU:** Fatora $A$ em $LU$, **preservando intacto o vetor $b$**, ideal para resolver múltiplos sistemas com a mesma matriz.
* **Gauss-Jordan:** Zera termos acima e abaixo da diagonal ($Ax=b \to Ix=c$), fornecendo o vetor solução sem retro-substituição, porém com custo de ~50% mais operações.

```python
# EXERCÍCIO DE FIXAÇÃO 4
A_fix4 = np.array([[1.0, 2.0, 1.0],
                   [3.0, 8.0, 7.0],
                   [2.0, 7.0, 9.0]])
b_fix4 = np.array([4.0, 20.0, 23.0])

# a) Matriz estendida inicial
Ab_fix4 = np.c_[A_fix4, b_fix4]
print("a) Matriz estendida inicial [A|b]:\n", Ab_fix4)

# b) Eliminação da coluna 1
Ab_fix4[1] -= (Ab_fix4[1, 0] / Ab_fix4[0, 0]) * Ab_fix4[0]
Ab_fix4[2] -= (Ab_fix4[2, 0] / Ab_fix4[0, 0]) * Ab_fix4[0]
print("\nb) Matriz intermediária A|b^(1) após primeira etapa:\n", Ab_fix4)
```

## Exercício de Revisão 8

**Por que operações elementares preservam rigorosamente o conjunto solução?**
1. **Reversibilidade Algébrica:** Cada operação elementar de linha possui inversa direta única da mesma natureza.
2. **Invertibilidade Matricial:** Cada operação equivale a pré-multiplicar o sistema por matrizes elementares inversíveis ($E_i$), de determinante não nulo:  
   $$Ax = b \iff (E_k \cdots E_1) Ax = (E_k \cdots E_1) b \iff M A x = M b$$
Como $M$ é inversível, os sistemas são logicamente equivalentes.

```python
# PROBLEMA 9: Gerenciamento de Energia em SoC (PMIC)
A_pmic = np.array([[ 15.0, -5.0, -2.0],
                   [ -5.0, 18.0, -6.0],
                   [ -2.0, -6.0, 10.0]])
b_pmic = np.array([12.0, 0.0, 5.0])

print("PROBLEMA 9")
# a) Matriz estendida
Ab_pmic = np.c_[A_pmic.copy(), b_pmic.copy()]
print("a) Matriz estendida inicial:\n", Ab_pmic)

# b) Escalonamento passo a passo
m21 = Ab_pmic[1, 0] / Ab_pmic[0, 0]
m31 = Ab_pmic[2, 0] / Ab_pmic[0, 0]
print(f"\nb) Multiplicadores coluna 1: m21 = {m21:.4f}, m31 = {m31:.4f}")
Ab_pmic[1] -= m21 * Ab_pmic[0]
Ab_pmic[2] -= m31 * Ab_pmic[0]

m32 = Ab_pmic[2, 1] / Ab_pmic[1, 1]
print(f"   Multiplicador coluna 2: m32 : {m32:.4f}")
Ab_pmic[2] -= m32 * Ab_pmic[1]
print("   Matriz triangular escalonada final:\n", np.round(Ab_pmic, 4))

# c) Tensões nodais
V = substituicao_retroativa(Ab_pmic[:, :-1], Ab_pmic[:, -1])
print(f"\nc) Tensões nodais:\n   V1 = {V[0]:.4f} V\n   V2 = {V[1]:.4f} V\n   V3 = {V[2]:.4f} V")
```

```python
# PROBLEMA 12: Cluster de Edge Computing
A_edge = np.array([[ 1.0, -0.1, -0.2],
                   [-0.2,  1.0, -0.1],
                   [-0.1, -0.3,  1.0]])
b_edge = np.array([100.0, 150.0, 200.0])

print("PROBLEMA 12")
Ab_edge = np.c_[A_edge.copy(), b_edge.copy()]
n = len(A_edge)
pivos = []
for c in range(n - 1):
    pivos.append(Ab_edge[c, c])
    for l in range(c + 1, n):
        Ab_edge[l] -= (Ab_edge[l, c] / Ab_edge[c, c]) * Ab_edge[c]
pivos.append(Ab_edge[-1, -2])
print("a) Pivôs observados:", [round(p, 4) for p in pivos])
print("   Todos não nulos?", all(not np.isclose(p, 0.0) for p in pivos))
print("\nb) Sistema triangular U:\n", np.round(Ab_edge[:, :-1], 4))

x_edge = substituicao_retroativa(Ab_edge[:, :-1], Ab_edge[:, -1])
print("\nc) Taxas de processamento calculadas:")
for i, taxa in enumerate(x_edge, 1):
    print(f"   Nó S{i}: {taxa:.4f} req/s")
```

```python
# PROBLEMA 14: Contagem de FLOPs e Análise de Complexidade
def escalona_com_flops(A, b):
    Ab = np.c_[A.astype(float), b.astype(float)]
    n = len(A)
    flops = 0
    for c_idx in range(n - 1):
        for l_idx in range(c_idx + 1, n):
            m = Ab[l_idx, c_idx] / Ab[c_idx, c_idx]
            flops += 1
            num_elem = (n + 1) - (c_idx + 1)
            Ab[l_idx, c_idx + 1:] -= m * Ab[c_idx, c_idx + 1:]
            Ab[l_idx, c_idx] = 0.0
            flops += num_elem
    return Ab[:, :-1], Ab[:, -1], flops

print("PROBLEMA 14")
dimensoes = [10, 50, 100, 200, 400]
print(f"{'n':<6} | {'FLOPs Medidos':<15} | {'Teórico (n^3 / 3)':<25}")
print("-" * 55)
for n_dim in dimensoes:
    A_rand = np.random.rand(n_dim, n_dim) + np.eye(n_dim) * 10
    b_rand = np.random.rand(n_dim)
    _, _, flops = escalona_com_flops(A_rand, b_rand)
    print(f"{n_dim:<6} | {flops:<15} | {(n_dim**3)/3.0:<25.1f}")

print("\na) A eliminação domina com complexidade O(n^3) (3 laços aninhados), enquanto a retro-substituição é O(n^2).")
print("b) Dobrando o tamanho de n=1000 para n=2000, o tempo aumenta por um fator de 2^3 = 8 vezes.")
```
