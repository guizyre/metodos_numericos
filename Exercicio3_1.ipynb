```python
import matplotlib.pyplot as plt
import numpy as np

v = np.array([1, 2])
w = np.array([4, -6])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))


ax1.quiver(
    0, 0, v[0], v[1], angles="xy", scale_units="xy", scale=1, color="black"
)

ax1.quiver(
    v[0],
    v[1],
    w[0],
    w[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    color="dimgray",
)

v_plus_w = v + w
ax1.quiver(
    0,
    0,
    v_plus_w[0],
    v_plus_w[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    color="lightgray",
)

ax1.set_xlim([-6, 6])
ax1.set_ylim([-6, 6])
ax1.set_xticks(np.arange(-6, 7, 2))
ax1.set_yticks(np.arange(-6, 7, 2))
ax1.grid(True, linestyle="-", alpha=0.5)
ax1.set_aspect("equal")
ax1.set_title("A)\nVectors v, w, and v+w", fontweight="bold")
ax1.legend(["v", "w", "v+w"], loc="upper right")


ax2.quiver(
    0, 0, v[0], v[1], angles="xy", scale_units="xy", scale=1, color="black"
)

ax2.quiver(
    0, 0, w[0], w[1], angles="xy", scale_units="xy", scale=1, color="dimgray"
)

v_minus_w = v - w
ax2.quiver(
    w[0],
    w[1],
    v_minus_w[0],
    v_minus_w[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    color="lightgray",
)

ax2.set_xlim([-6, 6])
ax2.set_ylim([-6, 6])
ax2.set_xticks(np.arange(-6, 7, 2))
ax2.set_yticks(np.arange(-6, 7, 2))
ax2.grid(True, linestyle="-", alpha=0.5)
ax2.set_aspect("equal")
ax2.set_title("B)\nVectors v, w, and v-w", fontweight="bold")
ax2.legend(["v", "w", "v-w"], loc="upper right")

plt.tight_layout()
plt.show()
```

```python
import numpy as np


def compute_norm(v):
    v_flat = v.ravel()

    sum_squares = 0.0
    for i in range(len(v_flat)):
        sum_squares += v_flat[i] ** 2

    return np.sqrt(sum_squares)


np.random.seed(42)

test_vectors = [
    np.random.randn(3),
    np.random.randn(15),
    np.random.randn(50, 1),
    np.random.randn(1, 100),
]

for i, v in enumerate(test_vectors, start=1):
    my_result = compute_norm(v)
    np_result = np.linalg.norm(v)

    match = np.isclose(my_result, np_result)
    print(
        f"Teste {i} {v.shape}: Algoritmo = {my_result:.6f} | np.linalg.norm = {np_result:.6f} | Igual: {match}"
    )
```

```python
import numpy as np
# Exercício 2-3

def make_unit_vector(v):
    norm = np.linalg.norm(v)
    if norm == 0:
        print("Aviso (Ex 2-3): Vetor nulo detetado. Não é possível divisão por zero.")
        return v
    return v / norm

print("--- Exercício 2-3 ---")
v1 = np.array([3, 4])
unit_v1 = make_unit_vector(v1)
print(f"Vetor unitário de {v1}: {unit_v1} | Norma: {np.linalg.norm(unit_v1)}")

v_zeros = np.array([0, 0, 0])
unit_zeros = make_unit_vector(v_zeros)
print(f"Resultado para vetor nulo: {unit_zeros}\n")

def set_vector_magnitude(v, desired_magnitude):
    norm = np.linalg.norm(v)
    if norm == 0:
        return v
    unit_vector = v / norm
    return unit_vector * desired_magnitude

print("--- Exercício 2-4 ---")
v2 = np.array([1, 1, 1])
desired_mag = 5
v2_new_mag = set_vector_magnitude(v2, desired_mag)
print(f"Vetor original: {v2}")
print(f"Novo vetor com magnitude {desired_mag}: {v2_new_mag}")
print(f"Nova norma confirmada: {np.linalg.norm(v2_new_mag):.2f}\n")



print("--- Exercício 2-5 ---")
# Vetor linha (1 x 4)
v_row = np.array([[10, 20, 30, 40]])
print(f"Vetor Linha original:\n{v_row} | Shape: {v_row.shape}")

n_cols = v_row.shape[1]

v_col = np.zeros((n_cols, 1))

for i in range(n_cols):
    v_col[i, 0] = v_row[0, i]

print(f"Vetor Coluna transposto manualmente:\n{v_col} | Shape: {v_col.shape}\n")



# Exercício 2-6
print("Exercício 2-6 ")
v3 = np.array([2, -5, 4, 1])

dot_product = np.dot(v3, v3)


squared_norm = np.linalg.norm(v3)**2

print(f"Produto escalar (v . v): {dot_product}")
print(f"Quadrado da norma (||v||^2): {squared_norm}")
print(f"São equivalentes? {np.isclose(dot_product, squared_norm)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

a = np.array([2, 5])
b = np.array([5, 2])


beta = np.dot(a, b) / np.dot(a, a)

proj_p = beta * a

fig, ax = plt.subplots(figsize=(6, 6))

ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)

ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='black', width=0.007)
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='black', width=0.007)

ax.plot([proj_p[0], b[0]], [proj_p[1], b[1]], 'k--', linewidth=1.5)

ax.plot(proj_p[0], proj_p[1], 'wo', markeredgecolor='black', markersize=10, markeredgewidth=1.5)

ax.text(a[0] + 0.1, a[1] + 0.1, 'a', fontsize=14, fontweight='bold')
ax.text(b[0] + 0.1, b[1] + 0.1, 'b', fontsize=14, fontweight='bold')
ax.text(proj_p[0] - 0.4, proj_p[1] + 0.2, r'$\beta$', fontsize=14, fontweight='bold')
ax.text((b[0] + proj_p[0])/2 + 0.2, (b[1] + proj_p[1])/2 + 0.2, r'(b - $\beta$a)', fontsize=12, fontweight='bold')

ax.set_xticks([])
ax.set_yticks([])

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
t = np.array([0.75, -0.8]) + np.random.randn(2) * 0.1
r = np.array([1.3, -0.15]) + np.random.randn(2) * 0.1

beta = np.dot(t, r) / np.dot(r, r)
t_par = beta * r

t_perp = t - t_par



sum_components = t_par + t_perp
is_sum_correct = np.allclose(sum_components, t)

dot_product = np.dot(t_par, t_perp)
is_orthogonal = np.isclose(dot_product, 0.0)

print(f"Os componentes somam 't'? {is_sum_correct}")
print(f"Produto escalar entre t_|| e t_perp: {dot_product:.2e}")
print(f"Os componentes são perfeitamente ortogonais? {is_orthogonal}\n")

#Figura 2-8

fig, ax = plt.subplots(figsize=(6, 6))

ax.plot([0, t[0]], [0, t[1]], 'k-', linewidth=3, label='t')
ax.plot([0, r[0]], [0, r[1]], color='silver', linewidth=3, label='r')
ax.plot([0, t_par[0]], [0, t_par[1]], 'k--', linewidth=3, label=r'$\mathbf{t}_{||}$')
ax.plot([0, t_perp[0]], [0, t_perp[1]], 'k:', linewidth=3, label=r'$\mathbf{t}_{\perp}$')

ax.legend(loc='upper right', prop={'weight': 'bold', 'size': 12})
ax.set_aspect('equal')
ax.set_xlim([-0.2, 1.5])
ax.set_ylim([-1.2, 0.4])

plt.show()
```

```python
import numpy as np

np.random.seed(42)
t = np.random.randn(2)
r = np.random.randn(2)

# (Equação 2-13)

beta_correct = np.dot(t, r) / np.dot(r, r)
t_par_correct = beta_correct * r
t_perp_correct = t - t_par_correct


beta_bug = np.dot(t, r) / np.dot(t, t)
t_par_bug = beta_bug * r
t_perp_bug = t - t_par_bug

dot_check_correct = np.dot(t_perp_correct, r)
dot_check_bug = np.dot(t_perp_bug, r)

print("--- Verificação de Ortogonalidade ---")
print(f"Produto escalar correto (esperado ~0): {dot_check_correct:.2e}")
print(f"Produto escalar com bug (esperado ~0): {dot_check_bug:.2e}")

is_correct_orthogonal = np.isclose(dot_check_correct, 0)
is_bug_orthogonal = np.isclose(dot_check_bug, 0)

print(f"\nA versão correta é ortogonal? {is_correct_orthogonal}")
print(f"A versão com bug é ortogonal? {is_bug_orthogonal}")
```

```python
import numpy as np

escalares = [2, -1, 3]
vetores = [
    np.array([1, 2, 3]),
    np.array([4, 5, 6]),
    np.array([7, 8, 9])
]

resultado_direto = escalares[0]*vetores[0] + escalares[1]*vetores[1] + escalares[2]*vetores[2]

resultado_loop = np.zeros(len(vetores[0]))

for i in range(len(escalares)):
    resultado_loop += escalares[i] * vetores[i]

print(f"Resultado sem ciclo: {resultado_direto}")
print(f"Resultado com ciclo: {resultado_loop}")
print(f"Os resultados são iguais? {np.allclose(resultado_direto, resultado_loop)}")
```
