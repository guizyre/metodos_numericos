### Resolução do Exercício de Fixação 1

A **eliminação gaussiana convencional** (sem estratégias de pivoteamento) pode apresentar sérios problemas computacionais nas seguintes situações:

1. **Divisão por Zero (Pivô Nulo):**
   - **Situação:** Se, durante o processo de eliminação, o elemento na diagonal principal (o pivô $a_{kk}$) for exatamente zero, a operação de divisão para calcular os multiplicadores ($m_{ik} = a_{ik}/a_{kk}$) se torna impossível.
   - **Problema:** O algoritmo é interrompido por um erro de divisão por zero, mesmo que o sistema de equações lineares possua uma solução única e bem definida.

2. **Instabilidade Numérica (Pivô Muito Próximo de Zero):**
   - **Situação:** Se o pivô $a_{kk}$ for um valor muito pequeno em magnitude em comparação com os outros elementos da coluna.
   - **Problema (Cancelamento Catastrófico e Erros de Arredondamento):** Dividir por um número muito pequeno gera multiplicadores extremamente grandes. Ao somar esses multiplicadores gigantes aos termos restantes da matriz usando aritmética de ponto flutuante (que tem precisão finita), os números de menor magnitude originais são "apagados" ou severamente distorcidos. Isso propaga erros de arredondamento gigantescos, tornando a solução final completamente incorreta.

### Resolução do Exercício de Fixação 2

No contexto de sistemas lineares parametrizados por um valor $\epsilon$ (comumente associados a problemas de quase-singularidade ou perturbações):

- **Comportamento quando $\epsilon \to \infty$:**
  - Quando o parâmetro $\epsilon$ cresce indefinidamente, os termos que dependem diretamente dele passam a dominar completamente a matriz de coeficientes do sistema.
  - Fisicamente ou geometricamente, isso geralmente faz com que as equações representadas no sistema se tornem redundantes ou paralelas (se houver dependência linear assintótica), ou faz com que os coeficientes associados a certas variáveis fiquem infinitamente maiores que os outros.
  - Do ponto de vista numérico, o número de condicionamento do sistema tende a se deteriorar severamente, ou as variáveis tornam-se completamente desacopladas de forma trivial onde certas incógnitas tendem a zero para manter a consistência física/matemática do sistema original.

### Resolução do Exercício de Fixação 3

Ambas as abordagens apresentadas falham devido ao mesmo problema lógico fundamental de perda de informação por sobrescrita imediata.   

#### 1. Análise do Código com `for`:

* **O que acontece:**
  Na linha `A[k, j] = A[p, j]`, o valor original de `A[k, j]` é sobrescrito e perdido.
  Na linha seguinte, `A[p, j] = A[k, j]`, é atribuido o valor que agora está em `A[k, j]` de volta para `A[p, j]`.
* Resultado: No final de cada iteração, ambas as posições `A[k, j]` e `A[p, j]` terão exatamente o mesmo valor original de `A[p, j]`. Os dados originais da linha `k` foram apagados.

#### 2. Análise do Código com Fatiamento (Slicing):

* **O que acontece:**
  * No NumPy, fatiamentos slices como `A[p, k:]` criam visões da memória original, e não cópias físicas imediatas por padrão.
  * Ao executar `A[k, k:] = A[p, k:]`, os valores da linha `k` são imediatamente atualizados com os da linha `p`.
  * Na instrução seguinte, como `A[k, k:]` foi alterado na memória, a atribuição `A[p, k:] = A[k, k:]` apenas copia os dados já modificados de volta para `p`.



### Exercício de Revisão 4:

1. **Inviabilidade Algorítmica:**
   Quando o elemento pivô ($a_{kk}$) assume valor nulo em alguma etapa da eliminação, o cálculo dos multiplicadores de linha torna-se matematicamente impossível por configurar uma divisão por zero. O pivotamento resolve essa limitação por meio do rearranjo das equações, substituindo o pivô nulo por um coeficiente adequado.

2. **Instabilidade Numérica:**
   Quando o valor do pivô é excessivamente próximo de zero em relação aos demais coeficientes da coluna, o cálculo dos multiplicadores resulta em valores com ordens de grandeza muito elevadas. Sob aritmética de ponto flutuante, a operação subsequente com ordens de magnitude discrepantes provoca erros graves por cancelamento catastrófico e perda de precisão, comprometendo a exatidão do resultado final. O pivotamento busca o maior elemento em magnitude para atuar como pivô, garantindo a estabilidade numérica do processo.

Exercício de Revisão 5:

Professor, de forma bem resumida, a diferença está em onde nós procuramos o melhor elemento para ser o pivô:

* **Pivotamento Parcial:** A busca é feita apenas na mesma coluna do pivô atual. Nós só trocamos linhas de lugar. É a implementação mais comum porque é muito rápida computacionalmente e já resolve a imensa maioria dos problemas de instabilidade.
* **Pivotamento Completo:** A busca é muito mais rigorosa. Nós procuramos o maior valor em toda a submatriz restante. Por causa disso, exige trocar linhas e colunas de lugar. É muito mais estável numericamente, mas custa caro computacionalmente e raramente compensa o esforço extra na prática.

# **Problema 7**

```python
def escalona_pivot_parcial(A, b):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)
    Ab = np.c_[A, b]
    for c in range(n - 1):
        p = np.abs(Ab[c:, c]).argmax() + c
        if p != c:
            Ab[[c, p]] = Ab[[p, c]]
        for l in range(c + 1, n):
            m = Ab[l, c] / Ab[c, c]
            Ab[l] = Ab[l] - m * Ab[c]
    A_ = Ab[:, :-1]
    b_ = Ab[:, -1]
    return A_, b_


A_ex1 = np.array([[1, 1, 1], [4, 4, 2],
                  [2, 1, -1]])
b_ex1 = np.array([1, 2, 0])

A_esc, b_esc = escalona_pivot_parcial(A_ex1, b_ex1)
print("Matriz A escalonada:\n", A_esc)
print("\nVetor b escalonado:\n", b_esc)
```

# Problema **8**

```python
def escalona_com_roll(A, b):
    Ab = np.c_[np.array(A, dtype=float), np.array(b, dtype=float)]
    n = len(b)
    for c in range(n - 1):
        p = np.abs(Ab[c:, c]).argmax() + c
        if p != c:
            shift = -(p - c)
            Ab[c : p + 1] = np.roll(Ab[c : p + 1], shift=shift, axis=0)
        for l in range(c + 1, n):
            m = Ab[l, c] / Ab[c, c]
            Ab[l] = Ab[l] - m * Ab[c]
    return Ab[:, :-1], Ab[:, -1]

A_esc_roll, b_esc_roll = escalona_com_roll(A_ex1, b_ex1)
print("Escalonamento usando np.roll:\n", A_esc_roll)
print("\nVetor b escalonado:\n", b_esc_roll)
```
