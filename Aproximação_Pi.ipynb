```python
import numpy as np
np.set_printoptions(precision=8,suppress=True)

termos = int(input("digite a quantidade de termos: "));
valor_final = 0.0;

for c in range(termos):
  termo = ((-1)**c) / (2*c + 1);
  valor_final += termo;
  print(4 * valor_final);
```
