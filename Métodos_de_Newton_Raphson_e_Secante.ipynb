```python
import numpy as np

def f1(x): return 2*x**4 + 4*x**3 + 3*x**2 - 10*x - 15
def f2(x): return (x + 3)*(x + 1)*(x - 2)**3
def f3(x): return 5*x**3 + x**2 - np.exp(1 - 2*x) + np.cos(x) + 20
def f4(x): return np.sin(x)*x + 4
def f5(x): return ((x - 3)**5) * np.log(x)
def f6(x): return x**10 - 1




```

# **método de Newton-Raphson**

```python
import numpy as np

def new_raph(f,x0,tol,max_iter, h=1e-7):
  x = x0
  for k in range(max_iter):
    fx = f(x)
    dfx = (f(x + h) - f(x - h)) / (2 * h)

    if abs(f(x)) < tol:
        print(f'A raiz encontrada na interação {k+1} é {x}')
        return x
    if dfx == 0:
      print("Erro! Derivada igual 0")
      return None
    x -= (fx/dfx)
  print(f"Erro de convergência: limite de {max_iter} iterações atingido.")
  return None
```

```python
raiz1 = new_raph(f1, 1.5, 10**-10, 200)
raiz2 = new_raph(f2, 2.5, 10**-10, 200)
raiz3 = new_raph(f3, 0, 10**-10, 200)
raiz4 = new_raph(f4, 3, 10**-10, 200)
raiz5 = new_raph(f5, 3.5, 10**-10, 200)
raiz5 = new_raph(f6, 1, 10**-10, 200)
```

# **Método Secante**

```python
import numpy as np

def secante(f, x0, x1, tol, max_iter):
    fx0 = f(x0)
    fx1 = f(x1)

    for k in range(max_iter):
        if fx1 - fx0 == 0:
            print("Erro! divisão por zero detectada")
            return None

        x2 = x1 - fx1 * (x1 - x0) / (fx1 - fx0)
        fx2 = f(x2)

        if abs(fx2) < tol:
            print(f"A raiz encontrada na iteração {k+1} é: {x2}")
            return x2

        x0 = x1
        fx0 = fx1

        x1 = x2
        fx1 = fx2

    print(f"Erro de convergência: limite de {max_iter} iterações atingido.")
    return None
```

```python
raiz1 = secante(f1,0,3,10**-10,200)
raiz2 = secante(f2,0,5,10**-10,200)
raiz3 = secante(f3,-5,5,10**-10,200)
raiz4 = secante(f4,1,5,10**-10,200)
raiz5 = secante(f5,2,5,10**-10,200)
raiz6 = secante(f6,0.8,1.2,10**-10,200)
```
