Questão 1


```python
import numpy as np

tolerancia = 0.000001
i = 0;
a = int(input("Digite o valor de a: "))
b = int(input("Digite o valor de b: "))

def funcoes(x):
  return (6+x) * (8-x)

func_a = funcoes(a)
func_b = funcoes(b)

produto_sinais = np.sign(func_a) * np.sign(func_b)

if produto_sinais > 0:
  print("Não existe garantia de raiz neste intervalo sinais iguais. \nPrograma Encerrado")
  exit(1)

elif produto_sinais == 0:
  print("Um dos extremos digitados já é uma raiz exata!")
  if func_a == 0:
    print(f"Raiz encontrada exatamente em: {a}")
  if func_b == 0:
    print(f"Raiz encontrada exatamente em: {b}")

else:
  while (abs(b - a) > tolerancia):
    ponto_medio = (a + b) / 2

    func_pm = funcoes(ponto_medio)
    print(f"Iteração {i:02d} | a = {a:.6f} | b = {b:.6f} | Ponto Médio = {ponto_medio:.6f} | f(pm) = {func_pm:.6f}")

    if func_pm == 0:
      break

    if np.sign(func_a) * np.sign(func_pm) == -1:
      b = ponto_medio
      func_b = func_pm
    else:
      a = ponto_medio
      func_a = func_pm

    i += 1

  print(f"Raiz encontrada: {ponto_medio:.6f} após {i} iterações")
```

```python
import numpy as np
def f(x):
    return np.sin(x) - x**3

a = 0.5
b = 1.0
tolerancia = 0.02
i = 0
erro = 1.0
ponto_medio_anterior = 0

if np.sign(f(a)) * np.sign(f(b)) != -1:
    print("Não existe garantia de raiz no intervalo.")
    exit()
else:
    print(f"{'Iter':<5} | {'a':<8} | {'b':<8} | {'Ponto Médio':<12} | {'Erro (%)':<8}")

    while True:
        i += 1
        ponto_medio = (a + b) / 2
        func_pm = f(ponto_medio)

        if i > 1:
            erro = abs((ponto_medio - ponto_medio_anterior) / ponto_medio)

        erro_str = f"{erro * 100:.2f}%" if i > 1 else "-"
        print(f"{i:<5} | {a:<8.5f} | {b:<8.5f} | {ponto_medio:<12.5f} | {erro_str:<8}")

        if func_pm == 0 or (i > 1 and erro < tolerancia):
            break

        if np.sign(f(a)) * np.sign(func_pm) == -1:
            b = ponto_medio
        else:
            a = ponto_medio

        ponto_medio_anterior = ponto_medio

    print("-" * 55)
    print(f"Raiz aproximada: {ponto_medio:.5f} atingida na iteração {i}")
```

```python
import numpy as np

def f_linha(x):
    return -12 * (x**5) - 6 * (x**3) + 10

a = 0.0
b = 1.0
tolerancia = 0.05
i = 0
erro = 1.0
ponto_medio_anterior = 0

if np.sign(f_linha(a)) * np.sign(f_linha(b)) != -1:
    print("Não existe garantia de raiz no intervalo.")
    exit()
else:
    print(f"{'Iter':<5} | {'a':<8} | {'b':<8} | {'Ponto Médio':<12} | {'Erro (%)':<8}")
    while True:
        i += 1

        ponto_medio = (a + b) / 2
        func_pm = f_linha(ponto_medio)

        if i > 1:
            if ponto_medio != 0:
                erro = abs((ponto_medio - ponto_medio_anterior) / ponto_medio)
            else:
                erro = 0.0

        erro_str = f"{erro * 100:.2f}%" if i > 1 else "-"
        print(f"{i:<5} | {a:<8.5f} | {b:<8.5f} | {ponto_medio:<12.5f} | {erro_str:<8}")

        if func_pm == 0 or (i > 1 and erro < tolerancia):
            break

        if np.sign(f_linha(a)) * np.sign(func_pm) == -1:
            b = ponto_medio
        else:
            a = ponto_medio

        ponto_medio_anterior = ponto_medio
    print(f"Ponto de máximo aproximado: x = {ponto_medio:.5f} (atingido na iteração {i})")
```

```python
import numpy as np

def bissecao(f, xl, xu, tol=1e-6, max_iter=100):
    if np.sign(f(xl)) * np.sign(f(xu)) >= 0:
        return None, 0

    x_ant = xl
    for iteracao in range(1, max_iter + 1):
        x_m = (xl + xu) / 2.0
        fm = f(x_m)

        if abs(x_m - x_ant) < tol or abs(fm) < 1e-12:
            return x_m, iteracao

        if np.sign(f(xl)) * np.sign(fm) < 0:
            xu = x_m
        else:
            xl = x_m

        x_ant = x_m

    return x_m, max_iter


def falsa_posicao(f, xl, xu, tol=1e-6, max_iter=100):
    if np.sign(f(xl)) * np.sign(f(xu)) >= 0:
        return None, 0

    x_ant = xl
    for iteracao in range(1, max_iter + 1):
        fl = f(xl)
        fu = f(xu)

        x_star = xu - (fu * (xl - xu)) / (fl - fu)
        f_star = f(x_star)

        if abs(x_star - x_ant) < tol or abs(f_star) < 1e-12:
            return x_star, iteracao

        if np.sign(fl) * np.sign(f_star) < 0:
            xu = x_star
        else:
            xl = x_star

        x_ant = x_star

    return x_star, max_iter

"""comparação"""
def func_1(x):
    return np.sin(x) - x**3

def func_2(x):
    return x**3 - 2*x - 5

testes = [
    {"nome": "f(x) = sin(x) - x^3", "f": func_1, "xl": 0.5, "xu": 1.0},
    {"nome": "f(x) = x^3 - 2x - 5", "f": func_2, "xl": 2.0, "xu": 3.0}
]

print(f"{'Função':<22} | {'Método':<15} | {'Raiz Encontrada':<18} | {'Iterações'}")

for t in testes:
    raiz_b, iter_b = bissecao(t["f"], t["xl"], t["xu"])
    raiz_fp, iter_fp = falsa_posicao(t["f"], t["xl"], t["xu"])

    print(f"{t['nome']:<22} | {'Bisseção':<15} | {raiz_b:<18.6f} | {iter_b}")
    print(f"{'':<22} | {'Falsa Posição':<15} | {raiz_fp:<18.6f} | {iter_fp}")
```
